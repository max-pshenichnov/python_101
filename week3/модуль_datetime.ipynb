## Python модуль datetime

```python
from datetime import datetime, date, timedelta, timezone
```

```python
# текущее время
now = datetime.now()
print(now)
type(now)
```

```python
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
print(now.microsecond)
```

```python
# только дата
today = date.today()
print(today)
```

```python
# создать конкретную дату
lesson = datetime(2026, 9, 24, 19, 00)
print(lesson)
```

```python
# только день
day = datetime(2026, 9, 24)
print(day)
```

```python
# сравнение дат
a = datetime(2026, 1, 1)
b = datetime(2026, 9, 24)

print(a > b)
print(a < b)
print(a != b)
```

```python
delta = b - a

print(type(delta))
print(delta.days)
```

```python
three_days = timedelta(days=3)
future = datetime.now() + three_days
print(future)
```

```python
# строка из "времени" - разное оформление
text = future.strftime("%d.%m.%Y")
print(text)

text = future.strftime("%d | %m | %y")
print(text)

text = future.strftime("%m_%d")
print(text)
```

```python
my_text_date = "24.09.2026"
d = datetime.strptime(my_text_date, "%d.%m.%Y") # формат должен совпадать с форматом строки!
print(d)
```

```python
day = datetime.fromtimestamp(timestamp)
print(day)
```

## UTC и часовые пояса

```python
now_utc = datetime.now(timezone.utc)
print(now_utc)
```

```python
from zoneinfo import ZoneInfo
now_moscow = datetime.now(ZoneInfo("Europe/Moscow"))
print(now_moscow)
```
