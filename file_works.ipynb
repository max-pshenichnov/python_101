```python
def select_film(year_start, year_end):
    film_selection = []
    for film in films:
        try:
            year = int(film["year"])
        except ValueError:
            #print(f"ОШИБКА ДАННЫХ: {film['title']}")
            continue
        if year >= year_start and year <= year_end:
            #print(f"подходид: {film['title']}")
            film_selection.append(film)
    return film_selection
```

```python
selection = select_film(1950, 1955)
```

```python
from random import choice
```

```python
our_random_film = choice(selection)
```

```python
print(f'''сегодня смотрим {our_random_film['title_clean']}, {our_random_film['year']}, идёт {our_random_film['runtime']}''')
```
